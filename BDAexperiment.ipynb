APPLY K-MEANS CLUSTERING ON A DATASET AND EVALUATE THE CLUSTERING PERFORMANCE USING METRICS LIKE INERTIA AND SILHOUETTE SCORE

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import load_iris
from sklearn.metrics import silhouette_score

# 1. Load the Dataset
# For unsupervised learning, we only use the feature data (X), not the labels (y)
iris = load_iris()
X = iris.data

# 2. Apply K-Means Clustering
# We specify n_clusters=3 because the Iris dataset naturally has 3 classes
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans.fit(X)

# Extract cluster labels and centroids
cluster_labels = kmeans.labels_
centroids = kmeans.cluster_centers_

# 3. Evaluate Clustering Performance
# Calculate Inertia
inertia = kmeans.inertia_

# Calculate Silhouette Score
sil_score = silhouette_score(X, cluster_labels)

# Print the evaluation metrics
print("--- Evaluation Metrics ---")
print(f"Inertia: {inertia:.2f}")
print(f"Silhouette Score: {sil_score:.4f}")

# 4. Visualize the Clusters
# Plotting the first two features (Sepal Length vs Sepal Width) for 2D visualization
plt.figure(figsize=(8, 6))

# Scatter plot of the data points, colored by their assigned cluster
plt.scatter(X[:, 0], X[:, 1], c=cluster_labels, cmap='viridis',
            marker='o', edgecolor='k', s=50, alpha=0.7, label='Data Points')

# Scatter plot of the centroids
plt.scatter(centroids[:, 0], centroids[:, 1],
            s=200, c='red', marker='X', label='Centroids')

plt.title('K-Means Clustering (Sepal Length vs Sepal Width)')
plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1])
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

TEXT MINING WITH PYTHON: USE LIBRARIES LIKE NLTK OR SPACY TO PROCESS TEXT DATA.

```python
import spacy

# Load the pre-trained English language model
nlp = spacy.load("en_core_web_sm")

# Define the raw text data to be processed
text_data = "Apple is looking at buying U.K. startup for $1 billion. Tim Cook announced this on October 19th."

# Process the text using the spaCy NLP pipeline
doc = nlp(text_data)

print("--- TEXT MINING: EXTRACTING ENTITIES ---")
# Extract and print named entities, their start/end positions, and their labels
for entity in doc.ents:
    print(f"Entity: {entity.text:15} | Label: {entity.label_}")

print("\n--- TEXT MINING: PART-OF-SPEECH TAGGING (First 5 words) ---")
# Extract grammatical properties for the first few words
for token in doc[:5]:
    print(f"Word: {token.text:10} | POS: {token.pos_} | Explanation: {spacy.explain(token.pos_)}")
```

PERFORM TASKS LIKE TOKENIZATION, STEMMING, AND STOPWORD REMOVAL

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

# Download required NLTK datasets
nltk.download('punkt_tab')
nltk.download('stopwords')

text = "The developers are coding rapidly because they love building amazing applications."

# 1. Tokenization
# Convert to lowercase and split into words
tokens = word_tokenize(text.lower())
print("--- 1. TOKENIZATION ---")
print(tokens, "\n")

# 2. Stopword Removal
# Load English stopwords
stop_words = set(stopwords.words('english'))

# Filter out the stopwords (and standard punctuation)
filtered_tokens = [word for word in tokens if word.isalnum() and word not in stop_words]
print("--- 2. STOPWORD REMOVAL ---")
print(filtered_tokens, "\n")

# 3. Stemming
# Initialize the Porter Stemmer
stemmer = PorterStemmer()

# Apply stemming to the filtered list
stemmed_tokens = [stemmer.stem(word) for word in filtered_tokens]
print("--- 3. STEMMING ---")
print(stemmed_tokens)
```

1. USE ARIMA MODELS TO FORECAST DATA AND ANALYZE TRENDS AND SEASONALITY IN TIME SERIES DATA.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.tsa.arima.model import ARIMA
import warnings
warnings.filterwarnings('ignore')

# 1. Generate Synthetic Time Series Data
# Simulating 5 years of monthly sales data with an upward trend and seasonality
np.random.seed(42)
dates = pd.date_range(start='2020-01-01', periods=60, freq='MS')

# Trend (increasing) + Seasonality (sine wave) + Random Noise
trend = np.linspace(10, 50, 60)
seasonality = 10 * np.sin(np.linspace(0, 3.14 * 10, 60))
noise = np.random.normal(0, 2, 60)

data = trend + seasonality + noise
df = pd.DataFrame({'Sales': data}, index=dates)

# 2. Analyze Trend and Seasonality (Decomposition)
print("--- 1. DECOMPOSING TIME SERIES ---")
print("Visualizing Observed, Trend, Seasonal, and Residual components...\n")
decomposition = seasonal_decompose(df['Sales'], model='additive', period=12)

# Plot the decomposition
fig = decomposition.plot()
fig.set_size_inches(10, 8)
plt.suptitle('Time Series Decomposition', fontsize=14)
plt.tight_layout()
plt.show()

# 3. Fit the ARIMA Model
# Order parameters (p, d, q). Here we use:
# p=5 (look back 5 periods), d=1 (1st order differencing), q=0
print("--- 2. FITTING ARIMA MODEL ---")
model = ARIMA(df['Sales'], order=(5, 1, 0))
model_fit = model.fit()

# Print the coefficient table
print(model_fit.summary().tables[1])

# 4. Forecast Future Data (Next 12 Months)
print("\n--- 3. FORECASTING FUTURE VALUES ---")
forecast_steps = 12
forecast = model_fit.forecast(steps=forecast_steps)

# Create a dataframe for the forecasted dates
forecast_dates = pd.date_range(start='2025-01-01', periods=forecast_steps, freq='MS')
forecast_series = pd.Series(forecast.values, index=forecast_dates)
print(forecast_series.head()) # Print first few forecasted values

# 5. Visualize Historical Data vs Forecast
plt.figure(figsize=(12, 6))
plt.plot(df.index, df['Sales'], label='Historical Sales', marker='o')
plt.plot(forecast_series.index, forecast_series, label='ARIMA Forecast', color='red', linestyle='--', marker='x')

plt.title('Time Series Forecasting using ARIMA Model')
plt.xlabel('Date')
plt.ylabel('Sales Volume')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
